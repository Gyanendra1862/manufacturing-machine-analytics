```python
sns.histplot(df['temperature_c'], kde=True)
plt.title('Temperature Distribution')
plt.xlabel('Temperature (°C)')
plt.show()
```

```python
sns.boxplot(x=df['temperature_c'])
plt.title('Temperature Boxplot')
plt.xlabel('Temperature (°C)')
plt.show()
```

```python
sns.barplot(
    data=df,
    x='machine_id',
    y='downtime_hours',
    estimator='sum'
)
plt.xticks(rotation=90)
plt.title('Total Downtime by Machine')
plt.xlabel('Machine')
plt.ylabel('Downtime (Hours)')
plt.show()
```

```python
sns.countplot(data=df, x='failure_type')
plt.xticks(rotation=45)
plt.title('Failure Type Distribution')
plt.xlabel('Failure Type')
plt.ylabel('Number of Failures')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='failure_flag',
    y='temperature_c'
)
plt.title('Temperature vs Machine Failure')
plt.xlabel('Failure (0 = No, 1 = Yes)')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='failure_flag',
    y='vibration_mm_s'
)
plt.title('Vibration vs Machine Failure')
plt.xlabel('Failure (0 = No, 1 = Yes)')
plt.ylabel('Vibration (mm/s)')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='temperature_c',
    y='vibration_mm_s',
    hue='failure_flag'
)
plt.title('Temperature vs Vibration')
plt.xlabel('Temperature (°C)')
plt.ylabel('Vibration (mm/s)')
plt.show()
```

```python
monthly = df.groupby('Month')['downtime_hours'].sum()

monthly.plot(kind='line')
plt.title('Monthly Downtime Trend')
plt.xlabel('Month')
plt.ylabel('Downtime (Hours)')
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_cols = [
    'temperature_c',
    'vibration_mm_s',
    'pressure_bar',
    'load_pct',
    'rpm',
    'downtime_hours',
    'production_units',
    'defect_units',
    'energy_kwh',
    'maintenance_cost_inr'
]
plt.figure(figsize=(12,5))

corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
